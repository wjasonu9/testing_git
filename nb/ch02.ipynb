Duck typing: "if it walks and quacks like a duck, then it is a duck." As an example, we can implement a check for whether an object is iterable by checking if it implements the iterator protocol, ie if it has an <i>\_\_iter__</i> 'magic method,' ie if we can successfully call the <i>iter</i> function on it

```python
def isiterable(obj):
    try:
        iter(obj)
        return True
    except TypeError: # not iterable
        return False
isiterable('q]j')
```

ASCII can encode just about all characters on the keyboard in 1 byte (256 possibilities, the most common are numbered 32-127). Unicode is a much expanded set of characters. UTF-8 is the most common encoding (and Python's default) to map unicode to bytes. <a href=https://docs.python.org/3/howto/unicode.html>More</a> info<br>
<i>bytes</i> is already available when you start python3 (no import needed)

```python
qx='café'
print(qx.encode('utf-8'))
print(qx.encode('latin1')) #another encoding
byte_obj=b'S\xc3\xa3o Paulo' #precede opening quote with b for a bytes object
byte_obj.decode('utf-8') #reverse process of str.encode()
```

```python
#this approximates R's charToRaw(qx)
qj=qx.encode().hex() #type str
print(qj,qx.encode('latin1').hex())
bytes.fromhex(qj).decode() #recover qx
```

The <i>datetime</i> <a href=https://www.kdnuggets.com/2019/06/how-use-datetime.html>module</a> provides 3 types for a fixed point in time: <i>datetime, date, time</i>. The 1st combines the 2nd and 3rd, and is most common

```python
from datetime import datetime, date, time
qdt = datetime(2011, 10, 29, 20, 30, 21)
xd=qdt.date() #extract date
jt=qdt.time() #extract time
qdt.day #dir() shows what else you can extract
datetime.combine(xd,jt) #combine date and time
#convert date to datetime if you do not already have a time object https://stackoverflow.com/questions/1937622
datetime.combine(xd,time()) #time() is 0:00 by default
```

An <i>if</i> block executes only when its condition is True. It can be followed by 0+ <i>elif</i> blocks and a catchall <i>else</i> block if all other conditions are False. Once a True condition is found, no other blocks are reached

```python
q=3
if q<3:
    print("low")
elif 3<=q<6:
    print("medium")
else:
    print("high")
```

A <i>for</i> loop iterates over a collection. Use <i>continue</i> to go to the next iteration, skipping the rest of the code in the block, and <i>break</i> to exit the loop altogether (with nested loops, <i>continue/ break</i> just apply to the current loop, not outer loops)

```python
for i in [1,2,3]:
    for j in [1,2,3]:
        if j > i:
            break
        print((i, j))
my_list=[3,'q]j]x]',6.3,8]
tot=0
for qj in my_list:
    if not isinstance(qj, (int,float)):
        continue
    tot+=qj
tot
```

A <i>while</i> loop repeats until its condition is False, or a <i>break</i> explicitly ends it

```python
qj=288
tot=0
while qj>1:
  tot+=qj
  if tot>538:
    break
  qj=qj//2
tot
```
